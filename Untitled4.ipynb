```python


# Cell 1: Imports and Environment Setup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for professional charts
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
print("Libraries imported successfully!")

```

```python
# Cell 2: Data Loading / Simulation
# (If you have 'supermarket_sales.csv' in Colab, you can read it via pd.read_csv('supermarket_sales.csv'))
np.random.seed(42)
n_rows = 1000

data = {
    'Invoice ID': [f"{np.random.randint(100,999)}-{np.random.randint(10,99)}-{np.random.randint(1000,9999)}" for _ in range(n_rows)],
    'Branch': np.random.choice(['A', 'B', 'C'], size=n_rows),
    'City': np.random.choice(['Yangon', 'Naypyitaw', 'Mandalay'], size=n_rows),
    'Customer type': np.random.choice(['Member', 'Normal'], size=n_rows),
    'Gender': np.random.choice(['Male', 'Female'], size=n_rows),
    'Product line': np.random.choice([
        'Electronic accessories', 'Fashion accessories', 'Food and beverages',
        'Health and beauty', 'Home and lifestyle', 'Sports and travel'
    ], size=n_rows),
    'Unit price': np.round(np.random.uniform(10, 100, n_rows), 2),
    'Quantity': np.random.randint(1, 10, n_rows),
    'Date': pd.to_datetime(np.random.choice(pd.date_range('2026-01-01', '2026-03-31'), n_rows)).strftime('%m/%d/%Y'),
    'Time': np.random.choice([f"{h:02d}:{m:02d}" for h in range(10, 21) for m in [0, 15, 30, 45]], n_rows),
    'Payment': np.random.choice(['Cash', 'Credit card', 'Ewallet'], size=n_rows),
    'Rating': np.round(np.random.uniform(4, 10, n_rows), 1)
}

df = pd.DataFrame(data)
df['Tax 5%'] = np.round(df['Unit price'] * df['Quantity'] * 0.05, 2)
df['Total'] = np.round((df['Unit price'] * df['Quantity']) + df['Tax 5%'], 2)
df['COGS'] = np.round(df['Unit price'] * df['Quantity'], 2)
df['Gross margin percentage'] = 4.761904762
df['Gross income'] = np.round(df['Total'] * 0.05, 2)

print("Dataset shape:", df.shape)
df.head()
```

```python
 #Cell 3: Data Preprocessing (Handling missing values, duplicates, and types)
print("Missing values before cleaning:\n", df.isnull().sum())

# Remove duplicate records if any
initial_rows = len(df)
df.drop_duplicates(inplace=True)
print(f"Removed {initial_rows - len(df)} duplicate rows.")

# Correct data types
df['Date'] = pd.to_datetime(df['Date'])
df['Customer type'] = df['Customer type'].astype('category')
df['Branch'] = df['Branch'].astype('category')
df['Product line'] = df['Product line'].astype('category')

# Export clean dataset required for deliverables[span_2](start_span)[span_2](end_span)
df.to_csv('cleaned_supermarket_sales.csv', index=False)
print("Cleaned dataset successfully saved as 'cleaned_supermarket_sales.csv'!")
```

```python
# Cell 4: Descriptive Statistics and Group-wise Analysis
print("--- Descriptive Statistics ---")
print(df.describe())

print("\n--- Group-wise Total Sales by Branch ---")
print(df.groupby('Branch')['Total'].sum())

print("\n--- Pivot Table: Average Sales by Customer Type & Payment Method ---")
pivot_table = pd.pivot_table(df, values='Total', index='Customer type', columns='Payment', aggfunc=np.mean)
print(pivot_table)
```

```python
# Cell 5: Data Visualization (Generating at least 10 visualizations)[span_4](start_span)[span_4](end_span)
fig, axes = plt.subplots(5, 2, figsize=(15, 25))
fig.suptitle('Supermarket Sales Analysis - Exploratory Visualizations', fontsize=18, fontweight='bold', y=0.92)

# 1. Distribution Plot: Customer Ratings
sns.histplot(df['Rating'], kde=True, ax=axes[0, 0], color='purple')
axes[0, 0].set_title('1. Distribution of Customer Ratings')

# 2. Count Plot: Product Line Frequency
sns.countplot(y=df['Product line'], order=df['Product line'].value_counts().index, ax=axes[0, 1], palette='viridis')
axes[0, 1].set_title('2. Frequency of Transactions by Product Line')

# 3. Bar Chart: Total Sales Revenue by Branch
branch_sales = df.groupby('Branch')['Total'].sum().reset_index()
sns.barplot(x='Branch', y='Total', data=branch_sales, ax=axes[1, 0], palette='magma')
axes[1, 0].set_title('3. Total Sales Revenue by Branch')

# 4. Grouped Bar Chart: Product Line Sales across Branches
sns.barplot(x='Product line', y='Total', hue='Branch', data=df, estimator=sum, ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('4. Sales by Product Line Across Branches')
axes[1, 1].tick_params(axis='x', rotation=45)

# 5. Box Plot: Unit Price distribution by Product Line
sns.boxplot(x='Unit price', y='Product line', data=df, ax=axes[2, 0], palette='coolwarm')
axes[2, 0].set_title('5. Unit Price Distribution Across Product Lines')

# 6. Count Plot: Payment Methods Share
sns.countplot(x='Payment', data=df, ax=axes[2, 1], palette='Set1')
axes[2, 1].set_title('6. Adoption Share of Payment Methods')

# 7. Count Plot: Customer Type by Gender
sns.countplot(x='Customer type', hue='Gender', data=df, ax=axes[3, 0], palette='Pastel1')
axes[3, 0].set_title('7. Customer Type Breakdown by Gender')

# 8. Line Plot: Daily Sales Trend over Time
daily_sales = df.groupby('Date')['Total'].sum().reset_index()
axes[3, 1].plot(daily_sales['Date'], daily_sales['Total'], color='teal', marker='o', linestyle='-')
axes[3, 1].set_title('8. Daily Sales Trend Over Time')
axes[3, 1].tick_params(axis='x', rotation=45)

# 9. Heatmap: Correlation Matrix
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='Blues', fmt='.2f', ax=axes[4, 0], cbar=False)
axes[4, 0].set_title('9. Correlation Heatmap of Numerical Features')

# 10. Scatter Plot: Gross Income vs Total Sales
sns.scatterplot(x='Total', y='Gross income', hue='Customer type', data=df, ax=axes[4, 1], palette='Set2')
axes[4, 1].set_title('10. Total Sales vs. Gross Income')

plt.tight_layout(rect=[0, 0.03, 1, 0.90])
plt.savefig('supermarket_eda_visualizations.png', dpi=300)
plt.show()

print("All 10 visualizations generated and saved successfully!")
```
